## 4. Расстояние до цели

Пусть цель находится в точке $g=(g_x,g_y)$. Для объекта $i$ с координатами $(x_i,y_i)$ расстояние до цели вычислим по формуле:

$$
d_i=\sqrt{(x_i-g_x)^2+(y_i-g_y)^2}.
$$

В нашем примере $g=(8,8)$, поэтому получаем:

$$
d_i=\sqrt{(x_i-8)^2+(y_i-8)^2}.
$$

NumPy может сразу вычислить расстояния для всех объектов. Номер ближайшего объекта выбирается так:

$$
i^*=\arg\min_i d_i.
$$

Знак $\arg\min$ означает: найти **номер элемента**, для которого значение минимально. В коде это делает `np.argmin(distances)`.

```python
target = np.array([8, 8])
print(target)
print(positions)
```

```python
target[0], positions[:, 0]
```

```python
dx = positions[:, 0] - target[0]

print(positions[:, 0].shape)
print(target[0].shape)

print(dx)

# https://numpy.org/doc/stable/user/basics.broadcasting.html
```

```python
target = np.array([8, 8])

dx = positions[:, 0] - target[0]
dy = positions[:, 1] - target[1]
print(dx.shape, dy.shape)
distances = np.sqrt(dx**2 + dy**2)
print("dists",distances.shape)

print("Расстояния до цели:", np.round(distances, 2))

nearest_id = np.argmin(distances)
print("Ближайший объект:", nearest_id)
print("Его координаты:", positions[nearest_id])
print("Расстояние:", round(distances[nearest_id], 2))
```

Покажем объекты и цель вместе. Звездой обозначена цель, красным - ближайший объект.

```python
plt.figure(figsize=(6, 5))

plt.scatter(positions[:, 0],
            positions[:, 1],
            s=80,
            color="royalblue",
            label="Объекты")
plt.scatter(target[0],
            target[1],
            s=220,
            marker="x", # v, ^, x, o, * ...
            color="orange", label="Цель")

plt.scatter(
    positions[nearest_id, 0],
    positions[nearest_id, 1],
    s=200,
    facecolors="none",  # закраска внутри
    edgecolors="red",   # цвет ребер
    linewidths=4,       # толщина границы
    label="Ближайший",
)

for i in range(len(positions)):
    plt.plot( [positions[i, 0], target[0]],
              [positions[i, 1], target[1]],
              linestyle = "--",
              color="gray",
              alpha=0.3, )

plt.title("Поиск ближайшего объекта")
plt.xlabel("x")
plt.ylabel("y")
plt.xlim(0, 10)
plt.ylim(0, 10)
plt.grid(alpha=0.3)
plt.legend(loc="upper left")
plt.show()
```

## 5. Один шаг к источнику сигнала

Пусть источник сигнала находится в точке $\mathbf{g}$. Объекты не знают эту позицию абсолютно точно - каждый из них получает измеренную позицию с небольшой случайной ошибкой:

$$
\tilde{\mathbf{g}}_i^{(t)}=\mathbf{g}+\boldsymbol{\varepsilon}_i^{(t)}.
$$

Здесь $\tilde{\mathbf{g}}_i^{(t)}$ - позиция источника, которую измерил объект $i$, а $\boldsymbol{\varepsilon}_i^{(t)}$ - небольшой шум измерения. В среднем шум равен нулю, поэтому измерения располагаются около истинной позиции.

Объект движется не прямо к истинной точке, а к полученной оценке:

$$
\mathbf{x}_i^{(t+1)}=\mathbf{x}_i^{(t)}+\alpha\left(\tilde{\mathbf{g}}_i^{(t)}-\mathbf{x}_i^{(t)}\right).
$$

$$
\mathbf{x}_i^{(t+1)}=\mathbf{x}_i^{(t)}+\alpha\left(\mathbf{g}+\boldsymbol{\varepsilon}_i^{(t)}
-\mathbf{x}_i^{(t)}\right).
$$



Здесь $\alpha$ - доля расстояния до измеренной позиции, которую объект проходит за один шаг. В коде $\alpha=0.25$. Параметр `noise_std` задает величину ошибки. При `noise_std = 0.0` все объекты видят истинную позицию без шума.

```python
start_positions = np.array([
    [1.0, 2.0],
    [2.0, 6.0],
    [4.0, 4.0],
    [7.0, 2.0],
    [8.0, 7.0],
])

target = np.array([8.0, 8.0])  # Истинная позиция источника сигнала / цель
noise_std = 0.35              # Небольшая ошибка измерения

rng_signal = np.random.default_rng(42)

measurement_noise = rng_signal.normal(0, noise_std, size=start_positions.shape)
measured_targets = target + measurement_noise     # что происходит # ~g_i
print(measurement_noise)
print(target.shape, measurement_noise.shape)
print(measured_targets)
```

```python
direction = measured_targets - start_positions
experiment_positions = start_positions + 0.25 * direction

print("Начальные положения:\n", start_positions)
print("Измеренные позиции источника:\n", np.round(measured_targets, 2))
print("Положения после шага:\n", experiment_positions)
```

```python
plt.figure(figsize=(6, 5))
plt.scatter(start_positions[:, 0], start_positions[:, 1], s=80, label="Было")
plt.scatter(experiment_positions[:, 0], experiment_positions[:, 1], s=80, label="Стало")
plt.scatter(target[0], target[1], s=220, marker="*", color="orange", label="Источник сигнала")
plt.scatter(measured_targets[:, 0], measured_targets[:, 1], s=80, marker="x", color="crimson", label="Измеренные позиции источника")

for i in range(len(start_positions)):
    plt.plot(
        [start_positions[i, 0], experiment_positions[i, 0]],
        [start_positions[i, 1], experiment_positions[i, 1]],
        color="gray",
        alpha=0.7,
    )

plt.title("Один шаг по зашумленным измерениям")
plt.xlabel("x")
plt.ylabel("y")
plt.xlim(0, 10)
plt.ylim(0, 10)
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```

## 6. Несколько шагов

Чтобы повторить действие несколько раз, используется цикл `for`.

В примере ниже одна и та же команда движения выполняется пять раз. После каждого шага положение становится новым исходным положением, а каждый объект снова измеряет позицию источника. Поэтому ошибка немного меняется от шага к шагу.

Правило движения на шаге $t$ имеет вид:

$$
\mathbf{x}_i^{(t+1)}=\mathbf{x}_i^{(t)}+\alpha\left(\mathbf{g}+\boldsymbol{\varepsilon}_i^{(t)}-\mathbf{x}_i^{(t)}\right).
$$

Если шума нет, после одного шага остается доля $(1-\alpha)$ предыдущего расстояния:

$$
d_{t+1}=(1-\alpha)d_t.
$$

При наличии шума это равенство выполняется только приблизительно (по матожиданию) траектории слегка отклоняются, но при небольшом шуме объекты все равно собираются около источника.

```python
moving_positions = start_positions.copy()
step = 0.25
noise_std = 0.35
rng_movement = np.random.default_rng(11)

history = [moving_positions.copy()]

for step_number in range(5):
    measured_targets_step = target + rng_movement.normal(0, noise_std, size=moving_positions.shape)
    moving_positions = moving_positions + step * (measured_targets_step - moving_positions)
    history.append(moving_positions.copy())

history = np.array(history)
print("Размер истории:", history.shape)
print("Итоговые положения:\n", np.round(moving_positions, 2))
```

```python
plt.figure(figsize=(6, 5))

for object_id in range(len(start_positions)):
    plt.plot(
        history[:, object_id, 0],
        history[:, object_id, 1],
        marker="o",
        markersize=4,
        alpha=0.8,
    )

plt.scatter(target[0], target[1], s=240, marker="*", color="orange", label="Источник сигнала")
plt.title(f"Движение за пять шагов, шум = {noise_std}")
plt.xlabel("x")
plt.ylabel("y")
plt.xlim(0, 10)
plt.ylim(0, 10)
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```

## 7. Ищем хорошую точку

Теперь точка цели заранее не отмечена на графике. Вместо этого для каждой точки можно вычислить число - ее **оценку**.

Рассмотрим простую оценку:

$$
q(x,y)=(x-6)^2+(y-3)^2.
$$

Эта оценка равна квадрату расстояния до точки $(6,3)$. Корень можно не вычислять, точка с наименьшим квадратом расстояния будет также ближайшей.

Для нескольких точек выбираем номер точки с минимальной оценкой:

$$
i^*=\arg\min_i q(x_i,y_i).
$$

Чем меньше $q$, тем лучше точка. Лучшая точка находится около $(6,3)$, но представим, что программа знает только способ вычисления оценки.

```python
rng = np.random.default_rng(10)
candidates = rng.uniform(0, 10, size=(20, 2))

x = candidates[:, 0]
y = candidates[:, 1]
scores = (x - 6)**2 + (y - 3)**2

best_id = np.argmin(scores)
best_point = candidates[best_id]

print("Номер лучшей точки:", best_id)
print("Лучшая точка:", np.round(best_point, 2))
print("Ее оценка:", round(scores[best_id], 2))
```

```python
plt.figure(figsize=(7, 5))
points = plt.scatter(
    candidates[:, 0],
    candidates[:, 1],
    c=scores,
    s=90,
    cmap="viridis_r",
)

plt.scatter(best_point[0], best_point[1], s=120, marker="*", color="red", label="Лучшая найденная")

plt.scatter(6, 3, s=220, marker="^", color="violet", label="Цель")

plt.colorbar(points, label="Оценка: меньше - лучше")
plt.title("Сравнение случайных точек")
plt.xlabel("x")
plt.ylabel("y")
plt.xlim(0, 10)
plt.ylim(0, 10)
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```

## 8. Небольшой групповой поиск

Объединим знакомые действия:

1. создаем несколько случайных точек;
2. вычисляем их оценки;
3. находим лучшую точку группы;
4. немного двигаем остальные точки в ее сторону;
5. добавляем небольшое случайное отклонение, чтобы группа могла исследовать пространство;
6. повторяем действия несколько раз.

На шаге $t$ сначала выбирается лучшая точка группы:

$$
i^*=\arg\min_i q\left(\mathbf{x}_i^{(t)}\right),
\qquad
\mathbf{x}_{best}^{(t)}=\mathbf{x}_{i^*}^{(t)}.
$$

Затем каждый объект изменяет положение по правилу:

$$
\mathbf{x}_i^{(t+1)}=\mathbf{x}_i^{(t)}+\alpha\left(\mathbf{x}_{best}^{(t)}-\mathbf{x}_i^{(t)}\right)+\boldsymbol{\varepsilon}_i^{(t)}.
$$

Здесь $\alpha$ задает движение к лучшей точке, а $\boldsymbol{\varepsilon}$ - небольшое случайное смещение. В коде им соответствуют `move_to_best` и `noise`.

Это не готовый алгоритм. Это упрощенная модель группового поиска.

```python
rng = np.random.default_rng(5)

number_of_objects = 12
number_of_steps = 20
move_to_best = 0.25
random_move = 0.5 # std шума

group = rng.uniform(0, 10, size=(number_of_objects, 2))
group_history = [group.copy()]
best_score_history = []

for step_number in range(number_of_steps):
    # 1. Оцениваем все текущие точки
    x = group[:, 0]
    y = group[:, 1]
    scores = (x - 6)**2 + (y - 3)**2    # идиентично силе сигнала из Задач вступительных

    # 2. Находим лучшую точку группы
    best_id = np.argmin(scores)
    best_point = group[best_id].copy()
    best_score_history.append(scores[best_id])

    # 3. Двигаемся к лучшей точке и немного отклоняемся случайно
    noise = rng.normal(0, random_move, size=group.shape)
    group = group + move_to_best * (best_point - group) + noise

    # 4. Не позволяем координатам выйти за границы от 0 до 10
    group = np.clip(group, 0, 10)
    group_history.append(group.copy())

group_history = np.array(group_history)
best_score_history = np.array(best_score_history)

print("Лучшая оценка в начале:", np.round(best_score_history[0], 3))
print("Лучшая оценка в конце:", np.round(best_score_history[-1], 3))
```

$$
d_i \leq R.
$$

### Задание 11. Движение к центру группы

Среднее положение $n$ объектов называется центром группы:

$$
\mathbf{c}=\frac{1}{n}\sum_{i=1}^{n}\mathbf{x}_i.
$$

Каждый объект может пройти долю $\beta$ расстояния к этому центру:

$$
\mathbf{x}_i^{new}=\mathbf{x}_i+\beta(\mathbf{c}-\mathbf{x}_i).
$$

1. Запустите пример с `beta = 0.3`.
2. Попробуйте `beta = 0.0`, `0.7` и `1.0`.
3. Объясните, что происходит при $\beta=1$.

```python
rng_center = np.random.default_rng(14)
group_before = rng_center.uniform(0, 10, size=(12, 2))
center = group_before.mean(axis=0)
beta = 0.3  # ИЗМЕНИТЕ: попробуйте 0.0, 0.7 и 1.0
group_after = group_before + beta * (center - group_before)

print("Центр группы:", np.round(center, 2))

plt.figure(figsize=(6, 5))
plt.scatter(group_before[:, 0], group_before[:, 1], s=75, color="royalblue", label="До движения")
plt.scatter(group_after[:, 0], group_after[:, 1], s=75, color="seagreen", label="После движения")
plt.scatter(center[0], center[1], s=230, marker="*", color="orange", label="Центр")
for i in range(len(group_before)):
    plt.plot(
        [group_before[i, 0], group_after[i, 0]],
        [group_before[i, 1], group_after[i, 1]],
        color="gray",
        alpha=0.5,
    )
plt.title(f"Движение к центру, beta = {beta}")
plt.xlabel("x")
plt.ylabel("y")
plt.xlim(0, 10)
plt.ylim(0, 10)
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```

### Задание 11. Движение с фиксированной длиной шага

В пунктах 5–6 объект проходит долю оставшегося расстояния. Теперь сделаем длину перемещения постоянной.

Сначала получим единичное направление (единичный вектор) на цель:

$$
\mathbf{u}_i=\frac{\mathbf{g}-\mathbf{x}_i}{\|\mathbf{g}-\mathbf{x}_i\|}.
$$

Затем выполним шаг длины $s$:

$$
\mathbf{x}_i^{new}=\mathbf{x}_i+s\mathbf{u}_i.
$$

Если до цели осталось меньше $s$, объект перемещается точно в цель и не перескакивает через нее.

1. Попробуйте `step_length = 0.5`, `1.0` и `2.0`.
2. Сравните этот результат с заданием 5.
3. Объясните, в каком варианте перемещение зависит от расстояния до цели.

```python
fixed_positions = np.array([
    [1.0, 2.0],
    [2.0, 6.0],
    [4.0, 4.0],
    [7.0, 2.0],
    [8.0, 7.0],
])
fixed_target = np.array([8.0, 8.0])
step_length = 6.0  # ИЗМЕНИТЕ: попробуйте 0.5 и 2.0

directions = fixed_target - fixed_positions
distances = np.sqrt(np.sum(directions**2, axis=1))
safe_distances = np.maximum(distances, 1e-12)
unit_directions = directions / safe_distances[:, None]
real_step = np.minimum(step_length, distances)
fixed_positions_new = fixed_positions + real_step[:, None] * unit_directions

plt.figure(figsize=(6, 5))
plt.scatter(fixed_positions[:, 0], fixed_positions[:, 1], s=80, label="Было")
plt.scatter(fixed_positions_new[:, 0], fixed_positions_new[:, 1], s=80, label="Стало")
plt.scatter(fixed_target[0], fixed_target[1], s=230, marker="*", color="orange", label="Цель")
for i in range(len(fixed_positions)):
    plt.plot(
        [fixed_positions[i, 0], fixed_positions_new[i, 0]],
        [fixed_positions[i, 1], fixed_positions_new[i, 1]],
        color="gray",
        alpha=0.6,
    )
plt.title(f"Фиксированная длина шага: {step_length}")
plt.xlabel("x")
plt.ylabel("y")
plt.xlim(0, 10)
plt.ylim(0, 10)
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```

## 12. Анимация движения группы

В завершение посмотрим на групповой поиск в движении. Сначала создадим историю положений, а затем последовательно покажем ее кадры.

Синие точки - объекты, красная звезда - область с наилучшей оценкой, а зеленое кольцо - лучшая точка группы на текущем кадре.

Выполните две следующие ячейки по порядку. Для отображения используется `FuncAnimation` из состава Matplotlib.

```python
# создаем генератор случайных чисел с фиксированным начальным значением
rng_animation = np.random.default_rng(12)

# задаем число объектов, шагов и параметры движения
animation_objects = 15
animation_steps = 30
animation_move_to_best = 0.25
animation_random_move = 0.25

# создаем начальные положения объектов на поле от 0 до 10
animation_group = rng_animation.uniform(0, 10, size=(animation_objects, 2))

# сохраняем начальное положение как первый кадр
animation_history = [animation_group.copy()]

# последовательно выполняем шаги группового поиска
for step_number in range(animation_steps):
    # извлекаем координаты всех объектов
    x = animation_group[:, 0]
    y = animation_group[:, 1]

    # вычисляем оценку каждой точки
    animation_scores = (x - 6)**2 + (y - 3)**2

    # находим объект с наименьшей оценкой
    animation_best_id = np.argmin(animation_scores)
    animation_best_point = animation_group[animation_best_id].copy()

    # добавляем небольшое случайное смещение для исследования пространства
    animation_noise = rng_animation.normal(0, animation_random_move, size=animation_group.shape)

    # перемещаем объекты к лучшей точке с учетом случайного смещения
    animation_group = (
        animation_group
        + animation_move_to_best * (animation_best_point - animation_group)
        + animation_noise
    )

    # ограничиваем координаты границами поля
    animation_group = np.clip(animation_group, 0, 10)

    # сохраняем новое положение как следующий кадр
    animation_history.append(animation_group.copy())

# превращаем список кадров в массив numpy
animation_history = np.array(animation_history)
print("Кадров подготовлено:", len(animation_history))
```

```python
# подключаем инструмент для построения анимации
from matplotlib.animation import FuncAnimation

# выбираем формат отображения анимации в ноутбуке
plt.rcParams["animation.html"] = "jshtml"

# создаем область для построения графика
fig, ax = plt.subplots(figsize=(7, 5))

# задаем границы, подписи и оформление координатной плоскости
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("Анимация группового поиска")
ax.grid(alpha=0.3)

# создаем графические элементы, которые будут обновляться на каждом кадре
moving_points = ax.scatter([], [], s=80, color="royalblue", label="Объекты")
best_ring = ax.scatter([], [], s=190, facecolors="none", edgecolors="seagreen", linewidths=2, label="Лучшая точка группы")
ax.scatter(6, 3, s=240, marker="*", color="red", label="Наилучшая область")
step_label = ax.text(0.3, 9.5, "")
ax.legend(loc="upper right")

# функция обновляет положение точек и подпись номера шага
def update_animation(frame):
    # берем положения объектов для текущего кадра
    current_positions = animation_history[frame]

    # повторно вычисляем оценки и находим лучшую точку кадра
    current_scores = (current_positions[:, 0] - 6)**2 + (current_positions[:, 1] - 3)**2
    current_best_id = np.argmin(current_scores)

    # передаем новые координаты графическим элементам
    moving_points.set_offsets(current_positions)
    best_ring.set_offsets(current_positions[[current_best_id]])
    step_label.set_text(f"Шаг: {frame}")
    return moving_points, best_ring, step_label

# собираем последовательность кадров в анимацию
group_animation = FuncAnimation(
    fig,
    update_animation,
    frames=len(animation_history),
    interval=250,
    repeat=True,
)

# закрываем статическую фигуру, чтобы осталась только анимация
plt.close(fig)
group_animation
```
